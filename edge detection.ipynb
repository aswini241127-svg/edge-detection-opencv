```python
#Import required Modules
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Load Original Image
image = cv2.imread("download (2).jpg") 
```

```python
#Convert to GrayScale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
# Apply Sobel operator
sobelx  = cv2.Sobel(src = gray_image, ddepth = cv2.CV_64F, dx = 1, dy = 0, ksize = 3) 
sobely  = cv2.Sobel(src = gray_image, ddepth = cv2.CV_64F, dx = 0, dy = 1, ksize = 3)

sobelx = cv2.Sobel(gray_image, cv2.CV_64F, 1, 0, ksize=3)  
sobely = cv2.Sobel(gray_image, cv2.CV_64F, 0, 1, ksize=3)  
sobel_combined = cv2.magnitude(sobelx, sobely) 
```

```python
#Display all images
plt.figure(figsize = (12, 16))
plt.subplot(321); plt.imshow(image[:,:,::-1]); plt.title('Original')
plt.subplot(322); plt.imshow(gray_image, cmap='gray');plt.title('Grayscale') 
plt.subplot(323); plt.imshow(sobelx);plt.title('Sobel-X Edge Map')
plt.subplot(324); plt.imshow(sobely);plt.title('Sobel-Y Edge Map');
```

```python
#Sobel Combined
plt.figure(figsize = (12, 16))
plt.axis('off'); plt.imshow(sobel_combined, cmap='gray' ); plt.title('sobel_combined ');
```

```python
#Apply Prewitt Operator
# Prewitt kernels
prewitt_x = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1]
], dtype=np.float32)

prewitt_y = np.array([
    [-1, -1, -1],
    [ 0,  0,  0],
    [ 1,  1,  1]
], dtype=np.float32)

# Apply Prewitt filters
prewittx = cv2.filter2D(gray_image, cv2.CV_64F, prewitt_x)
prewitty = cv2.filter2D(gray_image, cv2.CV_64F, prewitt_y)

# Combine X and Y gradients
prewitt_combined = cv2.magnitude(prewittx, prewitty)
```

```python
# Display original image and Prewitt results

plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title("GrayScale Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(prewittx, cmap='gray')
plt.title("Prewitt X")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(prewitty, cmap='gray')
plt.title("Prewitt Y")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(prewitt_combined, cmap='gray')
plt.title("Prewitt Combined")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Apply Robert Operator
roberts_x = np.array([
    [1, 0],
    [0, -1]
], dtype=np.float32)

roberts_y = np.array([
    [0, 1],
    [-1, 0]
], dtype=np.float32)

# Apply Roberts filters
robertsx = cv2.filter2D(gray_image, cv2.CV_64F, roberts_x)
robertsy = cv2.filter2D(gray_image, cv2.CV_64F, roberts_y)

# Combine X and Y gradients
roberts_combined = cv2.magnitude(robertsx, robertsy)
```

```python
plt.figure(figsize=(15, 10))

plt.subplot(2, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title("GrayScale Image")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(robertsx, cmap='gray')
plt.title("Robert X")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(robertsy, cmap='gray')
plt.title("Robert Y")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(roberts_combined, cmap='gray')
plt.title("Robert Combined")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
#Apply Laplacian Operator
laplacian = cv2.Laplacian(gray_image, cv2.CV_64F)
```

```python
plt.figure(figsize = (12, 16))
plt.subplot(121); plt.axis('off'); plt.imshow(gray_image, cmap='gray'); plt.title('GrayScale Image')

plt.subplot(122);plt.imshow(laplacian, cmap='gray');plt.axis('off'); plt.title('laplacian Operator');
```

```python
#Canny edge detection
edges = cv2.Canny(gray_image, threshold1 = 180, threshold2 = 200)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(gray_image, cmap='gray')
plt.title("Gray Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(edges, cmap='gray')
plt.title("Canny Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
